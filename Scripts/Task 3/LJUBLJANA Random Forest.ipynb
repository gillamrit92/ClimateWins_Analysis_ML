# 2. Data Wrangling

```python
# Create a list of the columns containing "LJUBLJANA" in their names

ljl_list = list(x for x in X.columns if x.find('LJUBLJANA') >=0)
ljl_list
```

```python
# Create a dataframe with those columns

df_ljl = X[ljl_list]
```

```python
df_ljl.head()
```

```python
# Reduce answers dataset to LJUBLJANA answers only

answers_ljl = answers['LJUBLJANA_pleasant_weather']
answers_ljl
```

```python
df_ljl.shape
```

```python
answers_ljl.shape
```

# 3. Reshaping the modeling

```python
X2 = df_ljl
```

```python
y2 = answers_ljl
```

```python
# Turn X2 and y2 from df to arrays

X = np.array(X2)
y = np.array(y2)
```

```python
X.shape
```

```python
y.shape
```

# 4.  Data Split

```python
# Split data into train and test sets

X_train, X_test, y_train, y_test = train_test_split(X,y,random_state = 42)
```

```python
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

# 5. Random Forest Model

```python
# Create a RF classifier
clf = RandomForestClassifier(n_estimators = 100)#, max_depth=5)  
  
# Training the model on the training dataset
# fit function is used to train the model using the training sets as parameters
clf.fit(X_train, y_train)
```

```python
# Perform predictions on the test dataset
y_pred = clf.predict(X_test)
  
# Use metrics module for accuracy calculation
print("Model Accuracy: ", metrics.accuracy_score(y_test, y_pred))
```

```python
# Class-names = {0:'Unpleasant Weather', 1:'Pleasant Weather'}

fig = plt.figure(figsize=(80,40))
plot_tree(clf.estimators_[5], fontsize = 20, feature_names=df_ljl.columns, class_names=['unpleasant', 'pleasant'], filled=True)
plt.show();
```

```python
fig.savefig(os.path.join(path, 'Visualizations', 'random_forest_Lubj.png'),bbox_inches='tight')
```
